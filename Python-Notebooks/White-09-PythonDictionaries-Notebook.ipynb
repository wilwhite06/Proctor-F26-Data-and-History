Uses a for loop to go through each person's information in the bellevue_people dictionary, checks whether their age is greater than 20, and prints the string "(Name) is more than 20 years old. She is (age)." if the if statement is true.

```python
for person in bellevue_people.values():
    if person['age'] > 20:
        name = person['name']
        age = person['age']
        print(f'{name} is more than 20 years old. She is {age}.')
```

Uses a for loop with .items() to go through the bellevue_people dictionary and prints each "key"-"value" pair.

```python
for person in bellevue_people.items():
    print(person)
```
