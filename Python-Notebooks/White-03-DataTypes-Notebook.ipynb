Defines a function named split_into_words that involves all text from the file. splits it into individual words, and returns the data of the variable split_words.
This makes the text easier to read and count words later.

```python
#All code has to be grouped together in order for it to work correctly.
def split_into_words(any_chunk_of_text):
    lowercase_text = any_chunk_of_text.lower()
    split_words = re.split("\W+", lowercase_text)
    return split_words
```
