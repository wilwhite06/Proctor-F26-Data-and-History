## Code Workthrough

Defines a function that converts text to lower case, splits it into individual words, and returns the data of the variable split_words.
This makes the text easier to read and count words later.

```python
def split_into_words(any_chunk_of_text):
    lowercase_text = any_chunk_of_text.lower()
    split_words = re.split("\W+", lowercase_text)
    return split_words
```

Sets the file path for the text, *The Yellow Wallpaper* by Charlotte Perkins Gilman and sets the number of desired words to 40. This tells the code what text to analyze and how many words it should look for.

```python
filepath_of_text = "../texts/literature/The-Yellow-Wallpaper_Charlotte-Perkins-Gilman.txt"
number_of_desired_words = 40
```

Sets the variable "stopwords" to a list of words the code is not looking for in the text; these are most common words and articles in the text. This is so the code can look for words that truly have meaning in *The Yellow Wallpaper*.

```python
stopwords = ['i', 'me', 'my', 'myself', 'we', 'our', 'ours', 'ourselves', 'you', 'your', 'yours',
 'yourself', 'yourselves', 'he', 'him', 'his', 'himself', 'she', 'her', 'hers',
 'herself', 'it', 'its', 'itself', 'they', 'them', 'their', 'theirs', 'themselves',
 'what', 'which', 'who', 'whom', 'this', 'that', 'these', 'those', 'am', 'is', 'are',
 'was', 'were', 'be', 'been', 'being', 'have', 'has', 'had', 'having', 'do', 'does',
 'did', 'doing', 'a', 'an', 'the', 'and', 'but', 'if', 'or', 'because', 'as', 'until',
 'while', 'of', 'at', 'by', 'for', 'with', 'about', 'against', 'between', 'into',
 'through', 'during', 'before', 'after', 'above', 'below', 'to', 'from', 'up', 'down',
 'in', 'out', 'on', 'off', 'over', 'under', 'again', 'further', 'then', 'once', 'here',
 'there', 'when', 'where', 'why', 'how', 'all', 'any', 'both', 'each', 'few', 'more',
 'most', 'other', 'some', 'such', 'no', 'nor', 'not', 'only', 'own', 'same', 'so',
 'than', 'too', 'very', 's', 't', 'can', 'will', 'just', 'don', 'should', 'now', 've', 'll', 'amp']
```

The variable "full_text" opens the text file of *The Yellow Wallpaper* for the code to read through the document.

```python
full_text = open(filepath_of_text, encoding="utf-8").read()
```

Splits the entire text into individual words, removes the stop words, and counts how many times each of the remaining and meaningful words appear in the text. Then, it tallies the 40 most meaningful words to see how often these words appear in the text.

```python
all_the_words = split_into_words(full_text)
meaningful_words = [word for word in all_the_words if word not in stopwords]
meaningful_words_tally = Counter(meaningful_words)
most_frequent_meaningful_words = meaningful_words_tally.most_common(number_of_desired_words)
```

Runs the variable most_frequent_meaningful_words and displays the most common and important words from *The Yellow Wallpaper* in order to analyze the frequency of these words.

```python
most_frequent_meaningful_words
```
