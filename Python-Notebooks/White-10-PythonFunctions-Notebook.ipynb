Defines a function named make_text_shouty that takes text as a parameter, converts it to uppercase, and returns the modified text.

```python
def make_text_shouty(text):
    shouty_text = text.upper()
    return shouty_text
```

Defines a function named make_text_shoutier that converts text to uppercase, adds three exclamation marks to the uppercase text, and returns the modified text.

```python
def make_text_shoutier(text):
    shouty_text = text.upper()
    shoutier_text = shouty_text + '!!!'
    return shoutier_text
```

```python
make_text_shoutier("I like tacos")
```
