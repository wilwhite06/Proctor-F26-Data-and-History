Uses a for loop to go through each item in the gender list and checks whether the value is "m" or "f". If the value is "m", it prints "Person is male." If the value is "f", it prints "Person is female." Otherwise, it prints "Person's gender is not known."

```python
for person in gender:
    if person == 'm':
        print("Person is male")
    elif person == 'f':
        print("Person is female")
    else:
        print("Person's gender is not known")
```
